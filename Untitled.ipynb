```python
# Missing data cleaning tasks
import pandas as pd

df_missing = pd.DataFrame({
    'A': [None, 2, 3, None, 5],
    'B': [10, None, None, 40, 50],
    'C': [100, None, 300, 400, None]
})

print("Original df_missing:")
print(df_missing)
print("\n" + "-"*50 + "\n")

# 1) Eliminate any row that contains at least one NaN value
no_nans_any = df_missing.dropna(how='any')
print("1) Rows with no NaN in any column:")
print(no_nans_any)
print("\n" + "-"*50 + "\n")

# 2) Eliminate rows that have NaN values across all columns
no_nans_all = df_missing.dropna(how='all')
print("2) Rows that are not all-NaN:")
print(no_nans_all)
print("\n" + "-"*50 + "\n")

# 3) Eliminate rows that have at least one NaN in columns A and C only
no_nans_A_C = df_missing.dropna(subset=['A', 'C'], how='any')
print("3) Rows with no NaN in columns A and C:")
print(no_nans_A_C)
print("\n" + "-"*50 + "\n")

# 4) Replace all NaN values with a specified value
filled_custom = df_missing.fillna(-1)
print("4) All NaNs replaced with -1:")
print(filled_custom)
print("\n" + "-"*50 + "\n")

# 5) Replace all NaN values in column B with the mean of non-NaN values in B
mean_B = df_missing['B'].mean()
filled_B_mean = df_missing.copy()
filled_B_mean['B'] = filled_B_mean['B'].fillna(mean_B)
print(f"5) Column B mean: {mean_B}\n")
print("NaNs in B replaced with mean:")
print(filled_B_mean)
```
